```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import binom
```

```python
# Read the matches dataset
df = pd.read_csv("matches.csv")

# Display first five records
df.head()
```

```python
# Display number of rows and columns
print("Number of rows and columns:", df.shape)

print("\nColumn names:")
print(df.columns.tolist())
```

```python
# Get unique teams from team1 and team2 columns
teams = pd.concat([
    df["team1"],
    df["team2"]
]).dropna().unique()

print("Teams available in the dataset:\n")

for team in sorted(teams):
    print(team)
```

```python
# Select a team from the dataset
team = sorted(teams)[0]

print("Selected Team:", team)
```

```python
# Select all matches played by the selected team
team_matches = df[
    (df["team1"] == team) |
    (df["team2"] == team)
]

print("Total matches played by", team, ":", len(team_matches))
```

```python
# Remove matches where winner is missing
team_matches = team_matches.dropna(subset=["winner"])

# Count the number of matches won by the selected team
wins = (team_matches["winner"] == team).sum()

# Total matches with a known winner
matches_played = len(team_matches)

print("Team:", team)
print("Matches played:", matches_played)
print("Matches won:", wins)
```

```python
# Probability of winning a match
p = wins / matches_played

print("Probability of success (p):", p)
print("Probability of success (%):", p * 100)
```
