```python
#Compare Mean and Median
mean_value = df["Total Cases"].mean()
median_value = df["Total Cases"].median()

if mean_value > median_value:
    print("Mean > Median")
    print("Data tends to be right skewed.")

elif mean_value < median_value:
    print("Mean < Median")
    print("Data tends to be left skewed.")

else:
    print("Mean = Median")
    print("Data is approximately symmetric.")
```
