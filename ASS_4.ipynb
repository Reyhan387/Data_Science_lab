```python
import pandas as pd 
import numpy as np
```

```python
# Load dataset
df = pd.read_csv("CityTemperatures.csv")

print("Dataset:")
print(df)
```

```python
print(df.describe())
```

```python
# Range
```

```python
range_a = df['City_A'].max() - df['City_A'].min()
range_b = df['City_B'].max() - df['City_B'].min()
print(range_a)
print(range_b)
```

```python
#variance
var_a = df['City_A'].var()
var_b = df['City_B'].var()
print(f"Variance of City A: {var_a}")
print(f"Variance of City B: {var_b}")
```

```python
mean_a = df['City_A'].mean()
mean_b = df['City_B'].mean()
print(f"mean_a: {mean_a}")
print(f"mean_b: {mean_b}")
```

```python
std_a = df["City_A"].std()
std_b = df["City_B"].std()

print("City A:", std_a)
print("City B:", std_b)
```

```python
cv_a = (std_a / mean_a) * 100
cv_b = (std_b / mean_b) * 100

print("CV City A:", cv_a, "%")
print("CV City B:", cv_b, "%")
```

```python
# 6. Interpretation

if std_a < std_b:
    print("City A has more consistent temperatures.")
else:
    print("City B has more consistent temperatures.")
```
