```python
import numpy as np


# Student marks

marks = np.array([78,85,90,67,88,92,76,81])


print("Student Marks:")
print(marks)


# Total marks

total = np.sum(marks)

print("\nTotal Marks:",total)


# Average marks

average = np.mean(marks)

print("Average Marks:",average)


# Maximum marks

highest = np.max(marks)

print("Highest Marks:",highest)


# Minimum marks

lowest = np.min(marks)

print("Lowest Marks:",lowest)


# Standard deviation

std = np.std(marks)

print("Standard Deviation:",std)


# Rank of student

rank=np.argmax(marks)

print("Student with Highest Marks is at Index:",rank)
```
